### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
ingreso_tienda1 = tienda['Precio'].sum()
ingreso_tienda2 = tienda2['Precio'].sum()
ingreso_tienda3 = tienda3['Precio'].sum()
ingreso_tienda4 = tienda4['Precio'].sum()

print(f"Ingreso total de la Tienda 1: S/. {ingreso_tienda1:.2f}")
print(f"Ingreso total de la Tienda 2: S/. {ingreso_tienda2:.2f}")
print(f"Ingreso total de la Tienda 3: S/. {ingreso_tienda3:.2f}")
print(f"Ingreso total de la Tienda 4: S/. {ingreso_tienda4:.2f}")
```

##-> Gráfica del análisis de facturación

```python
import matplotlib.pyplot as plt

ingresos = [
    tienda['Precio'].sum(),
    tienda2['Precio'].sum(),
    tienda3['Precio'].sum(),
    tienda4['Precio'].sum()
]

nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

plt.figure(figsize=(8, 5))
bars = plt.bar(nombres_tiendas, ingresos, color='orange')

max_index = ingresos.index(max(ingresos))
bars[max_index].set_color('green')

plt.title('Ingresos Totales por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Ingreso ($)')
plt.tight_layout()

for i, ingreso in enumerate(ingresos):
    plt.text(i, ingreso + max(ingresos)*0.01, f"S/. {ingreso:,.2f}", ha='center', va='bottom')

plt.show()
```

# 2. Ventas por categoría

```python
# Tienda 1
print("Productos vendidos por categoría - Tienda 1:")
categorias_tienda1 = tienda['Categoría del Producto'].value_counts().sort_values(ascending=False)
print(f"{categorias_tienda1.index[0]}: {categorias_tienda1.iloc[0]} -> CATEGORÍA MÁS POPULAR")
for i in range(1, len(categorias_tienda1)):
    print(f"{categorias_tienda1.index[i]}: {categorias_tienda1.iloc[i]}")
print()

# Tienda 2
print("Productos vendidos por categoría - Tienda 2:")
categorias_tienda2 = tienda2['Categoría del Producto'].value_counts().sort_values(ascending=False)
print(f"{categorias_tienda2.index[0]}: {categorias_tienda2.iloc[0]} -> CATEGORÍA MÁS POPULAR")
for i in range(1, len(categorias_tienda2)):
    print(f"{categorias_tienda2.index[i]}: {categorias_tienda2.iloc[i]}")
print()

# Tienda 3
print("Productos vendidos por categoría - Tienda 3:")
categorias_tienda3 = tienda3['Categoría del Producto'].value_counts().sort_values(ascending=False)
print(f"{categorias_tienda3.index[0]}: {categorias_tienda3.iloc[0]} -> CATEGORÍA MÁS POPULAR")
for i in range(1, len(categorias_tienda3)):
    print(f"{categorias_tienda3.index[i]}: {categorias_tienda3.iloc[i]}")
print()

# Tienda 4
print("Productos vendidos por categoría - Tienda 4:")
categorias_tienda4 = tienda4['Categoría del Producto'].value_counts().sort_values(ascending=False)
print(f"{categorias_tienda4.index[0]}: {categorias_tienda4.iloc[0]} -> CATEGORÍA MÁS POPULAR")
for i in range(1, len(categorias_tienda4)):
    print(f"{categorias_tienda4.index[i]}: {categorias_tienda4.iloc[i]}")
print()

```

## -> Gráfico de las ventas por categoría

```python
cat1 = tienda['Categoría del Producto'].value_counts()
cat2 = tienda2['Categoría del Producto'].value_counts()
cat3 = tienda3['Categoría del Producto'].value_counts()
cat4 = tienda4['Categoría del Producto'].value_counts()

categorias_df = pd.DataFrame({
    'Tienda 1': cat1,
    'Tienda 2': cat2,
    'Tienda 3': cat3,
    'Tienda 4': cat4
}).fillna(0).astype(int)

categorias_df.plot(kind='bar', figsize=(12, 6))

plt.title('Cantidad de productos vendidos por categoría en cada tienda')
plt.xlabel('Categoría')
plt.ylabel('Cantidad de productos vendidos')
plt.xticks(rotation=45)
plt.legend(title='Tiendas')
plt.tight_layout()
plt.show()


```

## Gráfico del producto más y menos vendido por tienda

```python
mas_vendidos = []
menos_vendidos = []
nombres_mas_vendidos = []
nombres_menos_vendidos = []

# Tienda 1
prod1 = tienda['Producto'].value_counts()
nombres_mas_vendidos.append(prod1.index[0])
mas_vendidos.append(prod1.iloc[0])
nombres_menos_vendidos.append(prod1.index[-1])
menos_vendidos.append(prod1.iloc[-1])

# Tienda 2
prod2 = tienda2['Producto'].value_counts()
nombres_mas_vendidos.append(prod2.index[0])
mas_vendidos.append(prod2.iloc[0])
nombres_menos_vendidos.append(prod2.index[-1])
menos_vendidos.append(prod2.iloc[-1])

# Tienda 3
prod3 = tienda3['Producto'].value_counts()
nombres_mas_vendidos.append(prod3.index[0])
mas_vendidos.append(prod3.iloc[0])
nombres_menos_vendidos.append(prod3.index[-1])
menos_vendidos.append(prod3.iloc[-1])

# Tienda 4
prod4 = tienda4['Producto'].value_counts()
nombres_mas_vendidos.append(prod4.index[0])
mas_vendidos.append(prod4.iloc[0])
nombres_menos_vendidos.append(prod4.index[-1])
menos_vendidos.append(prod4.iloc[-1])

tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
x = range(len(tiendas))

plt.figure(figsize=(12, 6))
bar1 = plt.bar([i - 0.2 for i in x], mas_vendidos, width=0.4, color='seagreen', label='Más vendido')
bar2 = plt.bar([i + 0.2 for i in x], menos_vendidos, width=0.4, color='indianred', label='Menos vendido')

for i in range(len(tiendas)):
    plt.text(i - 0.2, mas_vendidos[i] + 0.5, f"{nombres_mas_vendidos[i]}\n({mas_vendidos[i]})",
             ha='center', va='bottom', fontsize=8)
    plt.text(i + 0.2, menos_vendidos[i] + 0.5, f"{nombres_menos_vendidos[i]}\n({menos_vendidos[i]})",
             ha='center', va='bottom', fontsize=8)

plt.title('Productos más y menos vendidos por tienda')
plt.xlabel('Tienda')
plt.ylabel('Cantidad de ventas')
plt.xticks(x, tiendas)
plt.legend()
plt.tight_layout()
plt.show()

```

## Gráfico del costo de envío promedio por tienda

```python
envios_promedio = [
    tienda['Costo de envío'].mean(),
    tienda2['Costo de envío'].mean(),
    tienda3['Costo de envío'].mean(),
    tienda4['Costo de envío'].mean()
]

tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

indice_max = envios_promedio.index(max(envios_promedio))

colores = ['lightskyblue'] * 4
colores[indice_max] = 'red'

plt.figure(figsize=(8, 5))
bars = plt.bar(tiendas, envios_promedio, color=colores)

for i, valor in enumerate(envios_promedio):
    plt.text(i, valor + max(envios_promedio)*0.02, f"S/. {valor:.2f}", ha='center')

plt.title('Costo de envío promedio por tienda')
plt.ylabel('Costo de envío promedio (S/.)')
plt.ylim(0, max(envios_promedio) * 1.2)
plt.tight_layout()
plt.show()
```
